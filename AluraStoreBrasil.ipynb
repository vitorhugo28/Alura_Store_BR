### Importação dos dados



```python
import pandas as pd

lojas_dados = {
    "Loja1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}


for nome, url in lojas_dados.items():
    df_loja= pd.read_csv(url)

    linhas, colunas = df_loja.shape

    nulos = df_loja.isnull().sum().sum()

    print(f"--- Relatório: {nome} ---")
    display(df_loja.head(3))
    print(" ")
    display(df_loja.tail(3))
    print(" ")
    print(f"Dimensões: {linhas} linhas e {colunas} colunas")
    print(f"Dados em branco encontrados: {nulos}")
    print("-" * 200)
```

#1. Análise do faturamento



```python
import pandas as pd
import matplotlib.pyplot as plt


urls = list(lojas_dados.values())
nomes_lojas = list(lojas_dados.keys())


faturamentos = []


def calcular_faturamento(url_loja):
    loja = pd.read_csv(url_loja)
    faturamento = loja["Preço"].sum()
    return faturamento

print("Análise do faturamento:")
for i in range(len(lojas_dados)):
    faturamentos.append(calcular_faturamento(urls[i]))
    print(f"Faturamento da loja {i + 1}: R${faturamentos[i]:,.2f}")


faturamento_total = sum(faturamentos)
print("-"*40)
print(f"Faturamento total: R${faturamento_total:,.2f}")


```

```python

cores = ['#4e79a7', '#f28e2b', '#e15759', '#76b7b2']

pd.Series(faturamentos, index=nomes_lojas).plot(kind="bar", figsize=(8, 6), color= cores)


plt.ticklabel_format(style='plain', axis='y')

plt.title("Faturamento por Loja")
plt.ylabel("Valor em R$")
plt.xticks(rotation=0)
plt.show()
plt.savefig('faturamento_por_loja.png')
```

```python
plt.figure(figsize=(8, 8))
cores_pizza = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
explode = (0.05, 0.05, 0.05, 0.05)

total_faturamento = sum(faturamentos)
porcentagens = [(f / total_faturamento) * 100 for f in faturamentos]

plt.pie(faturamentos, labels=nomes_lojas, autopct='%1.1f%%', startangle=90, colors=cores_pizza, explode=explode, shadow=True)
plt.title('Participação de Cada Loja no Faturamento Total', fontsize=15, fontweight='bold')
plt.axis('equal')
plt.savefig('participacao_faturamento_pizza.png')
plt.show()
```

# 2. Vendas por Categoria

```python
import pandas as pd


vendas_categoria_loja = {}


def calcular_vendas_categoria(url_loja):
    loja = pd.read_csv(url_loja)
    venda_categoria = loja.groupby("Categoria do Produto")["Preço"].sum()
    return venda_categoria


for i in range(len(lojas_dados)):
   resultado = calcular_vendas_categoria(urls[i])
   vendas_categoria_loja[nomes_lojas[i]] = resultado


df_vendas_final = pd.DataFrame(vendas_categoria_loja)

print("Vendas por Categoria:")
display(df_vendas_final)


df_ranking_nomes = pd.DataFrame({
    loja: df_vendas_final[loja].sort_values(ascending=False).index
    for loja in df_vendas_final.columns
})
print(" ")
print("Ranking de Categorias mais populares por Loja:")
display(df_ranking_nomes)
```

```python
plt.figure()
df_vendas_final.sum(axis=1).sort_values().plot(kind='barh', color='teal')
plt.title('Volume de Vendas por Categoria (Geral)', fontsize=15, fontweight='bold')
plt.xlabel('Total em Vendas (R$)')
plt.ylabel('Categoria')
plt.tight_layout()
plt.savefig('vendas_categoria.png')
plt.show()
```

# 3. Média de Avaliação das Lojas


```python
import pandas as pd
from numpy import median


medias_lojas = {}


def media_avaliacao(url_loja):
  loja = pd.read_csv(url_loja)
  media = loja["Avaliação da compra"].mean().round(2)
  return media


for i in range(len(lojas_dados)):
    medias_lojas[nomes_lojas[i]] = media_avaliacao(urls[i])


df_medias_lojas = pd.DataFrame(medias_lojas, index=['Média'])

print("Média de Avaliação das Lojas:")
display(df_medias_lojas)

```

# 4. Produtos Mais e Menos Vendidos

```python
import pandas as pd

ranking_produtos = {}


def produtos_vendidos(url_loja):
    loja = pd.read_csv(url_loja)
    vendas = loja["Produto"].value_counts()
    top_3 = vendas.head(3).index.tolist()
    bottom_3 = vendas.tail(3).index.tolist()

    return top_3 + bottom_3


for i in range(len(lojas_dados)):
    ranking_produtos[nomes_lojas[i]] = produtos_vendidos(urls[i])


indices = ["1º Mais Vendido", "2º Mais Vendido", "3º Mais Vendido",
           "3º Menos Vendido", "2º Menos Vendido", "1º Menos Vendido"]

df_ranking_completo = pd.DataFrame(ranking_produtos, index=indices)

print("Produtos Mais e Menos Vendidos:")
display(df_ranking_completo)

```

# 5. Frete Médio por Loja

```python
import pandas as pd

media_frete_loja = {}

def media_frete (url_loja):
    loja = pd.read_csv(url_loja)
    valor_media_frete = loja["Frete"].mean().round(2)
    return valor_media_frete

for i in range(len(lojas_dados)):
    media_frete_loja[nomes_lojas[i]] = media_frete(urls[i])



df_frete_loja = pd.DataFrame(media_frete_loja, index=["Frete Médio"])

print("Frete Médio por Loja:")
display(df_frete_loja)
```
